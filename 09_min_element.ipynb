```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

# Find minimum element
min_element = matrix[0][0]

for i in range(3):
    for j in range(3):
        if matrix[i][j] < min_element:
            min_element = matrix[i][j]

print("Minimum Element =", min_element)
```
