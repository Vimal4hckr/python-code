```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

# Search element
key = int(input("Enter element to search: "))

found = False

for i in range(3):
    for j in range(3):
        if matrix[i][j] == key:
            print("Element found at Row", i, "Column", j)
            found = True

if found == False:
    print("Element not found")
```
