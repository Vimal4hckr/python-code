```python
matrix1 = []
matrix2 = []
result = []

print("Enter 9 elements for Matrix 1:")
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix1.append(row)

print("Enter 9 elements for Matrix 2:")
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix2.append(row)

# Matrix Subtraction
for i in range(3):
    row = []
    for j in range(3):
        row.append(matrix1[i][j] - matrix2[i][j])
    result.append(row)

print("Resultant Matrix:")
for i in range(3):
    for j in range(3):
        print(result[i][j], end=" ")
    print()
```
