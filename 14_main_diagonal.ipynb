```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

# Print main diagonal elements
print("Main Diagonal Elements:")

for i in range(3):
    print(matrix[i][i], end=" ")
```
