```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

# Total sum
total = 0
for i in range(3):
    for j in range(3):
        total += matrix[i][j]

print("Total Sum =", total)
```
