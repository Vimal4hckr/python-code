```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

print("Boundary Elements:")

for i in range(3):
    for j in range(3):
        if i == 0 or i == 2 or j == 0 or j == 2:
            print(matrix[i][j], end=" ")
    print()
```
