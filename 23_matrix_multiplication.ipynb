```python
matrix1 = []
matrix2 = []
result = []

print("Enter 9 elements for Matrix 1:")
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix1.append(row)

print("Enter 9 elements for Matrix 2:")
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix2.append(row)

# Initialize result matrix with zeros
for i in range(3):
    row = []
    for j in range(3):
        row.append(0)
    result.append(row)

# Matrix Multiplication
for i in range(3):
    for j in range(3):
        for k in range(3):
            result[i][j] += matrix1[i][k] * matrix2[k][j]

print("Resultant Matrix:")
for i in range(3):
    for j in range(3):
        print(result[i][j], end=" ")
    print()
```
