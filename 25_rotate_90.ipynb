```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

print("Matrix after 90° Clockwise Rotation:")

for i in range(3):
    for j in range(2, -1, -1):
        print(matrix[j][i], end=" ")
    print()
```
