```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

# Count odd numbers
count = 0

for i in range(3):
    for j in range(3):
        if matrix[i][j] % 2 != 0:
            count += 1

print("Number of Odd Elements =", count)
```
