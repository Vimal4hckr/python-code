```python
matrix = []

print("Enter 9 elements:")

# Input matrix
for i in range(3):
    row = []
    for j in range(3):
        num = int(input())
        row.append(num)
    matrix.append(row)

# Print upper triangle
print("Upper Triangle Elements:")

for i in range(3):
    for j in range(3):
        if j >= i:
            print(matrix[i][j], end=" ")
        else:
            print(" ", end=" ")
    print()
```
